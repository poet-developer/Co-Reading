#### 라벨러마다 동의한 감정에 대한 빈도 및 분포 계산

```python
# 감정 라벨 데이터를 리스트로 변환하는 함수
def labels_to_list(labels_str):
    if pd.isna(labels_str):
        return []
    return [label.strip() for label in labels_str.split(',')]

# 라벨 데이터를 리스트로 변환하여 새로 열을 추가
df['annotator_01_label_list'] = df['annotator_01'].apply(labels_to_list)
df['annotator_02_label_list'] = df['annotator_02'].apply(labels_to_list)
df['annotator_03_label_list'] = df['annotator_03'].apply(labels_to_list)
df['annotator_04_label_list'] = df['annotator_04'].apply(labels_to_list)
df['annotator_05_label_list'] = df['annotator_05'].apply(labels_to_list)
```

```python
def statics_to_list(df):
    all_emotions = sum(df, [])

    # 감정 빈도 계산
    emotion_counts = pd.Series(all_emotions).value_counts()

    # 감정 분포 계산 (비율)
    emotion_distribution = emotion_counts / emotion_counts.sum()

    # 결과를 데이터프레임으로 정리
    emotion_stats = pd.DataFrame({
        '감정': emotion_counts.index,
        '빈도': emotion_counts.values,
        '분포(비율)': emotion_distribution.values
    })

    return emotion_stats
```

```python
# 결과 출력
annotator01_statics = statics_to_list(df['annotator_01_label_list'])
annotator02_statics = statics_to_list(df['annotator_02_label_list'])
annotator03_statics = statics_to_list(df['annotator_03_label_list'])
annotator04_statics = statics_to_list(df['annotator_04_label_list'])
annotator05_statics = statics_to_list(df['annotator_05_label_list'])
```

```python
def draw_graph_with_plotly(statics, title):
    """
    감정 분포를 그래프로 표현하는 함수 (Plotly 사용).
    
    Parameters:
        statics (DataFrame): '감정'과 '분포(비율)' 컬럼을 포함한 데이터프레임.
        title (str): 그래프 제목.
    """
    fig = px.bar(
        statics,
        x='감정',
        y='분포(비율)',
        color='감정',
        title=title,
        labels={'감정': '감정', '분포(비율)': '분포(비율)'},
        template='plotly_white'
    )
    fig.update_layout(
        xaxis=dict(tickangle=45),
        title=dict(font=dict(size=20)),
        xaxis_title="감정",
        yaxis_title="분포(비율)"
    )
    fig.show()
```

### 여러 행에 대해 각 주석자 감정의 "분포(비율)"를 계산한 다음, 감정 별 총합하여 최종 분포

```python
from collections import defaultdict

class EmotionDistributionCalculator:
    def __init__(self, df, annotator_columns): # 초기설정
        self.df = df
        self.annotator_columns = annotator_columns

    def calculate_row_emotion_distribution(self, emotion_list): 
        """
        한 행의 감정 리스트를 받아 감정 분포를 계산
        :param emotion_list: ['기쁨', '슬픔', '기쁨'] 등과 같은 감정 라벨 리스트
        :return: 감정:비율의 딕셔너리 (예: {'기쁨': 0.67, '슬픔': 0.33})
        """
        if not emotion_list: # 빈 리스트면 빈 딕셔너리 반환
            return {} 
        
        total = len(emotion_list)  # 총 감정 개수
        counts = pd.Series(emotion_list).value_counts()  # 감정별 등장 횟수 세기
        distribution = counts / total  # 감정별 비율 계산
        return distribution.to_dict() # 딕셔너리 형태로 반환

    def calculate_annotator_distribution(self, annotator):
        """
        특정 주석자에 대한 전체 감정 분포를 계산
        :param annotator: 주석자 컬럼명 (예: 'annotator01')
        :return: 해당 주석자의 감정 분포를 담은 DataFrame
        """
        total_distribution = defaultdict(float)  # 감정별 누적 분포를 저장할 딕셔너리

        # 모든 행(텍스트)에 대해 반복
        for i in range(len(self.df)):
            row_dist = self.calculate_row_emotion_distribution(self.df[annotator].iloc[i]) # 각 행의 감정 비율 계산
            for emotion, value in row_dist.items():
                total_distribution[emotion] += value # 감정별 분포를 누적합

         # 딕셔너리를 DataFrame으로 변환
        annotator_df = pd.DataFrame(list(total_distribution.items()), columns=['감정', '총합 분포(비율)'])
        annotator_df = annotator_df.sort_values(by='총합 분포(비율)', ascending=False).reset_index(drop=True) # 분포 기준으로 내림차순 정렬

        # 백분율로 환산
        annotator_df = self.convert_to_percentage(annotator_df)

        # 주석자 이름 컬럼 추가
        annotator_df['주석자'] = annotator

        return annotator_df

    def convert_to_percentage(self, distribution_df):
        """
        감정 분포를 백분율로 변환
        :param distribution_df: '총합 분포(비율)' 컬럼이 포함된 DataFrame
        :return: '백분율(%)' 컬럼이 추가된 DataFrame
        """
        total_sum = distribution_df['총합 분포(비율)'].sum() # 전체 합
        distribution_df['백분율(%)'] = (distribution_df['총합 분포(비율)'] / total_sum) * 100  # 백분율 계산
        
        return distribution_df

    def calculate_all_distributions(self):
        """
        모든 주석자에 대한 감정 분포를 계산하여 하나의 DataFrame으로 통합
        :return: 모든 주석자의 감정 분포가 포함된 통합 DataFrame
        """
        all_distributions = []

        for annotator in self.annotator_columns:
            annotator_df = self.calculate_annotator_distribution(annotator) # 각 주석자별 감정 분포 계산
            all_distributions.append(annotator_df)

        # 여러 주석자의 결과를 하나의 DataFrame으로 결합
        final_df = pd.concat(all_distributions, ignore_index=True)
        return final_df
```

```python
# 주석자 컬럼명을 리스트로 정의
# 각 컬럼은 이미 감정 라벨 리스트로 전처리된 상태여야 함
annotator_columns = ['annotator_01_label_list', 'annotator_02_label_list', 'annotator_03_label_list', 'annotator_04_label_list', 'annotator_05_label_list']

# 감정 분포 계산 클래스의 인스턴스를 생성
# 입력: 전체 데이터프레임과 주석자 컬럼 목록
calculator = EmotionDistributionCalculator(df, annotator_columns)

# 모든 주석자의 감정 분포를 계산하여 하나의 통합 DataFrame 생성
final_df = calculator.calculate_all_distributions()
```

```python
def split_final_df_by_annotator(final_df):
    segmented_dfs = {} # 주석자별 DataFrame을 저장할 딕셔너리 생성
    unique_annotators = final_df['주석자'].unique()  # '주석자' 열에서 고유 주석자 목록 추출

    for annotator in unique_annotators:
        # 해당 주석자의 행만 필터링하고, 인덱스를 초기화하여 새로운 DataFrame 생성
        segmented_dfs[annotator] = final_df[final_df['주석자'] == annotator].reset_index(drop=True)

    return segmented_dfs # 주석자별로 나눈 DataFrame 딕셔너리 반환
```

```python
def plot_full_emotion_distribution(df, value_col='백분율(%)', title='KPoEM 전체 감정 백분율 분포'):
    """
    44가지 감정의 백분율 분포를 plotly로 가로 막대 그래프로 시각화 (얇은 막대 포함)

    Parameters:
    - df: '감정'과 백분율 컬럼을 포함한 DataFrame
    - value_col: 시각화 기준이 되는 값 (기본값: '백분율(%)')
    - title: 그래프 제목
    """
    df_sorted = df.sort_values(by=value_col, ascending=True) # 감정별 비율을 기준으로 오름차순 정렬

    fig = px.bar(
        df_sorted,
        x=value_col, # 백분율(%) 기준으로 정렬
        y='감정', # 감정 기준으로 정렬
        orientation='h', # 가로 막대 그래프
        text=df_sorted[value_col].round(2), # 소수점 2자리로 반올림
        title=title
    )

    fig.update_layout(
        xaxis_title=value_col,
        yaxis_title='감정',
        template='plotly_white',
        height=44 * 17,  # 감정 1개당 높이 20px 정도로 조정 → 880px
        margin=dict(l=100, r=30, t=60, b=30)
    )

    fig.update_traces(marker=dict(line=dict(width=0.5, color='gray')), textposition='outside')

    fig.show()
```

#### 5명중 3명 이상이 동의한 감정에 대한 빈도 및 분포 계산

```python
def get_labels_agreed_by_at_least_k(row, k=3):
    """
    각 행(row)에 대해, 최소 k명 이상이 동의한 감정만 추출

    Parameters:
    - row: annotator label list들이 있는 DataFrame row
    - k: 동의한 annotator 최소 수 (기본 2명)

    Returns:
    - 감정 리스트 중 k명 이상이 공통으로 선택한 감정 리스트
    """
    all_labels = (
        row['annotator_01_label_list'] +
        row['annotator_02_label_list'] +
        row['annotator_03_label_list'] +
        row['annotator_04_label_list'] +
        row['annotator_05_label_list']
    )
    counter = pd.Series(all_labels).value_counts() # 감정별 개수 세기
    return [label for label, count in counter.items() if count >= k] # k명 이상이 동의한 감정 리스트
```

```python
# 3명 이상 동의한 감정 리스트로 새 컬럼 생성
df['common_labels'] = df.apply(lambda row: get_labels_agreed_by_at_least_k(row, k=3), axis=1)

df[['common_labels']].head(10) # 3명 이상 동의한 감정 리스트 확인
```

#### Agreement (데이터 당 감정별 동의한 주석자 단위 통계)

```python
from collections import Counter

def calculate_agreement_levels(df, k_range=[1, 2, 3, 4, 5]):
    agreement_counts = {k: 0 for k in k_range} # k_range에 있는 각 k에 대해 동의자 수를 카운트하기 위한 딕셔너리

    for idx, row in df.iterrows(): # iterrows()를 사용하여 각 행을 반복
        all_labels = (
            row['annotator_01_label_list'] +
            row['annotator_02_label_list'] +
            row['annotator_03_label_list'] +
            row['annotator_04_label_list'] +
            row['annotator_05_label_list']
        )
        label_counter = Counter(all_labels) # 각 라벨의 개수를 세기 위한 Counter 객체 생성
        max_label_agreement = max(label_counter.values()) if label_counter else 0
        # max_label_agreement는 각 행에서 가장 많이 나타나는 라벨의 개수 (최대값을)

        for k in k_range:
            if max_label_agreement >= k: 
                agreement_counts[k] += 1 #해당 딕셔너리의 k에 대해 동의자 수를 증가시킴

    # 비율 계산
    total = len(df)
    agreement_stats = pd.DataFrame({
        '최소 동의자 수': list(agreement_counts.keys()),
        '텍스트 수': list(agreement_counts.values()),
        '전체 대비 비율 (%)': [round(v / total * 100, 2) for v in agreement_counts.values()]
    })

    return agreement_stats
```

```python
agreement_df = calculate_agreement_levels(df) 
print(agreement_df) #KOTE / x=1 : 50,000 (100%) / x=2:
```

#### texts labeled for NO EMOTION (주석자 별 '없음'에 라벨링한 데이터 수)

```python
# 라벨 컬럼
label_cols = ['annotator_01', 'annotator_02', 'annotator_03', 'annotator_04', 'annotator_05']

# # 각 annotator 라벨 컬럼을 리스트로 변환
# for col in label_cols:
#     df[f'{col}_label_list'] = df[col].apply(labels_to_list)
```

```python
# 각 행에서 '없음'을 라벨링한 주석자 수 계산
def count_none_label(row, target='없음'):
    return sum(target in row[f'{col}_label_list'] for col in label_cols)

df['none_label_count'] = df.apply(count_none_label, axis=1)

# 없음 라벨 주석자 수에 따른 분포 집계
none_label_distribution = df['none_label_count'].value_counts().sort_index()
none_label_table = pd.DataFrame({
    '없음 라벨 주석자 수': none_label_distribution.index,
    '텍스트 수': none_label_distribution.values
})

none_label_table['전체 대비 비율 (%)'] = round(none_label_table['텍스트 수'] / len(df) * 100, 2)
```
