```python
import requests
import pandas as pd

# Zenodo record ID (DOI에서 추출)
record_id = "21131097"  # 예: 1234567

url = f"https://zenodo.org/api/records/{record_id}"
response = requests.get(url).json()

# 파일 목록 확인
files = response["files"]

for file in files:
    print(file["key"], file["links"]["self"])

file_url = files[0]["links"]["self"]

df = pd.read_csv(file_url)
print(df.head())
```

## 데이터 통계

```python
print("총 데이터:", len(df))
print("고유 감정 수:", df["emotion_term"].nunique())
print("reference type 수:", df["reference_type"].nunique())
```

```python
ref_stats = df["reference_type"].value_counts().reset_index()
ref_stats.columns = ["reference_type", "count"]

ref_stats["ratio (%)"] = (ref_stats["count"] / len(df) * 100).round(2)

print(ref_stats)
```

```python
color_stats = df["color_code"].value_counts().reset_index()
color_stats.columns = ["color_code", "count"]

color_stats["ratio (%)"] = (color_stats["count"] / len(df) * 100).round(2)

print(color_stats)
```
